```python
# Lista das variáveis de valor das faturas
bill_features = [
    'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
    'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6'
]

# Lista das variáveis de valor dos pagamentos
pay_features = [
    'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
    'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6'
]

# Lista com todas as características financeiras
financial_features = bill_features + pay_features

print("Características de valor da fatura:")
print(bill_features)

print("\nCaracterísticas de valor do pagamento:")
print(pay_features)

print("\nTodas as características financeiras:")
print(financial_features)

```

```python
import pandas as pd

# Carregar o dataset
df = pd.read_excel('default_of_credit_card_clients__courseware_version_1_21_19.xls')

df[bill_features].describe()
```

```python
import matplotlib.pyplot as plt

df[bill_features].hist(
    bins=20,
    layout=(2, 3),
    figsize=(14, 8)
)

plt.suptitle('Distribuição dos Valores das Faturas', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
df[pay_features].describe()
```

```python
df[pay_features].hist(
    bins=20,
    layout=(2, 3),
    figsize=(14, 8),
    xrot=45
)

plt.suptitle('Distribuição dos Valores de Pagamento', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
pay_zero_mask = df[pay_features] == 0

print("Quantidade de pagamentos iguais a zero por variável:")
print(pay_zero_mask.sum())

print("\nTotal geral de pagamentos iguais a zero:")
print(pay_zero_mask.sum().sum())
```

```python
import numpy as np # Importa a biblioteca numpy

# Substituir valores iguais a zero por NaN
df_pay_non_zero = df[pay_features].replace(0, np.nan)

# Aplicar log10 a todos os valores
df_pay_log10 = df_pay_non_zero.apply(np.log10)

# Plotar histogramas
df_pay_log10.hist(
    bins=20,
    layout=(2, 3),
    figsize=(14, 8),
    xrot=45
)

plt.suptitle('Histograma dos Pagamentos (log10) - Valores Diferentes de Zero', fontsize=16)
plt.tight_layout()
plt.show()
```
